```python
from src.chunking import MarkdownChunker
from src.store import EmbeddingStore
from src.models import Document
from pathlib import Path
from src.embeddings import LocalEmbedder
import uuid


# 1. Đọc file
text = ""
for i in range(1, 6):
    path = Path(f"/home/logan/Documents/vinuni/lab7/data/data{i}.md")
    text += path.read_text(encoding="utf-8") + "\n"

# 2. Chunk
chunker = MarkdownChunker(max_chunk_size=500)
chunks = chunker.chunk(text)
print(f"✅ Tổng số chunks: {len(chunks)}")

# 3. Tạo Document objects
doc_id = path.stem  # "data1"
docs = [
    Document(
        id=str(uuid.uuid4()),
        content=chunk,
        metadata={
            "doc_id": doc_id,
            "source": str(path),
            "chunk_index": i,
        }
    )
    for i, chunk in enumerate(chunks)
]

# 4. Khởi tạo store và lưu vào Chroma
store = EmbeddingStore(
    collection_name="vinuni_docs",
    embedding_fn=LocalEmbedder(model_name="paraphrase-multilingual-MiniLM-L12-v2")
)
store.add_documents(docs)
print(f"✅ Đã lưu {store.get_collection_size()} chunks vào store")
```

```python

# 5. Test search
query = "điều kiện học bổng tài năng dành cho sinh viên mới là gì?"
results = store.search(query, top_k=5)

print(f"\n🔍 Kết quả tìm kiếm cho: '{query}'")
print("=" * 60)
for i, result in enumerate(results):
    print(f"\n--- Kết quả {i+1} (score: {result['score']:.4f}) ---")
    print(f"Chunk index: {result['metadata'].get('chunk_index')}")
    print(result["content"])
```
